```python
from helper.extract_helper import collect_excel_files
from helper.cleaner_helper import _cast_all, _clean_dtype, _clean_value

excel_files = collect_excel_files()
```

```python
try:
    selected_file_index = int(input(
        f"Enter the number of the Excel file to process. Choose from 1 to {len(excel_files)}"
    ))
    if not 1 <= selected_file_index <= len(excel_files):
        raise ValueError
except ValueError:
    print("Invalid Input. Please enter a valid number.")
    exit()

print(f"Excel file path: {excel_files[selected_file_index - 1]}")
```

```python
import pandas as pd

workbook = pd.read_excel(excel_files[selected_file_index - 1], sheet_name = None, header = None)

file_name = os.path.basename(excel_files[selected_file_index - 1])

print(f"Loaded workbook: {file_name}")
print(f"Worksheet count: {len(workbook)}")

for sheet_name in workbook.keys():
    print(f"- {sheet_name}")
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "June",
          "July", "Aug", "Sep", "Oct", "Nov", "Dec"]

sheet_lookup: dict = {
    sheet_name.strip().lower(): sheet_name
    for sheet_name in workbook.keys()
}

monthly_sheets: list[str] = []

other_sheets: list[str] =["AUM", "NET SALES"]

for month in months:
    possible_monthly_sheets: list[str] = [
        f"{month.lower()} revised",
        f"{month.lower()} updated",
        month.lower()
    ]

    for possible_sheet in possible_monthly_sheets:
        if possible_sheet in sheet_lookup:
            monthly_sheets.append(sheet_lookup[possible_sheet])
            break

print(monthly_sheets)
```

```python
from helper.extract_helper import extract_monthly_data
from helper.print_helper import printData


extracted_monthly_data: dict = {
    month: extract_monthly_data(workbook[month])
    for month in monthly_sheets
}

for month, data in extracted_monthly_data.items():
    print(f" {month}")
    printData(data)
```
